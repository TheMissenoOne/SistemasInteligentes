```python
#Collecting data

# Source - https://stackoverflow.com/a/5711095
# Posted by Vishal, modified by community. See post 'Timeline' for change history
# Retrieved 2026-09-29, License - CC BY-SA 4.0

link ="https://archive.ics.uci.edu/static/public/17/breast+cancer+wisconsin+diagnostic.zip"

# open url
resp = urlopen(link)
# read zipfile
zipfile = ZipFile(BytesIO(resp.read()))
# get the csv file name
fname = zipfile.namelist()[0]
# convert to pandas dateframe
data = pd.read_csv(zipfile.open(fname), dtype=object,  names=["ID", "Diagnosis", "radius1", "texture1", "perimeter1", "area1","smoothness1", "compactness1", "concavity1", "concave_points1"])
# close zipfile we don't need
zipfile.close()

print(data)
```
